```python
def describe(s):
    """
    this function will provide the descriptive statics about the given series.
    It will also convert the list,dictionary,tuple to series.
    """
    s=pd.Series(s)
    print("sum of values:"+str(s.sum()))
    print("product of valuse:"+str(s.product()))
    print("mean of values:"+str(s.mean()))
    print("standard of valuse:"+str(s.std()))
    print("variance:"+str(s.var()))
    print("median:"+str(s.median()))
```

```python
describe(s)
```

```python
def update_value(s,id,va):
    """
    It will update the series with the value and index
    """
    s=pd.Series(s)
    s[id]=va
    return s
```

```python
update_value(s=[11,13,15,16],id=10,va=18)
```
